```python
# ============================================================
# 3. Feature Engineering + Missing Values Handling
# - Create train_df_fe / test_df_fe
# - Fill missing values safely
# - Add interaction features (domain-inspired)
# ============================================================

TARGET_COL = "diagnosed_diabetes"
ID_COL = "id"

train_df_fe = train_df.copy()
test_df_fe = test_df.copy()

# ----------------------------
# Handle missing values
# ----------------------------
# Numeric: fill with median (computed on train only)
numeric_cols = train_df_fe.select_dtypes(include=["int64", "float64"]).columns.tolist()
numeric_cols.remove(ID_COL)  # keep id untouched
numeric_cols.remove(TARGET_COL)  # target is not a feature

for col in numeric_cols:
    med = train_df_fe[col].median()
    train_df_fe[col] = train_df_fe[col].fillna(med)
    test_df_fe[col] = test_df_fe[col].fillna(med)

# Categorical: fill with "Unknown"
categorical_cols = train_df_fe.select_dtypes(include=["object"]).columns.tolist()
for col in categorical_cols:
    train_df_fe[col] = train_df_fe[col].fillna("Unknown")
    test_df_fe[col] = test_df_fe[col].fillna("Unknown")

# Target: drop rows with missing target (rare but must be handled)
before_rows = train_df_fe.shape[0]
train_df_fe = train_df_fe.dropna(subset=[TARGET_COL])
after_rows = train_df_fe.shape[0]
print(f"Dropped {before_rows - after_rows} rows with missing target.")

# ----------------------------
# Feature engineering (interaction features)
# ----------------------------
def add_interaction_features(df: pd.DataFrame) -> pd.DataFrame:
    """Add medically motivated interaction features."""
    df = df.copy()

    df["age_bmi"] = df["age"] * df["bmi"]
    df["activity_bmi"] = df["physical_activity_minutes_per_week"] / (df["bmi"] + 1e-3)
    df["chol_ratio"] = df["cholesterol_total"] / (df["hdl_cholesterol"] + 1e-3)
    df["bp_ratio"] = df["systolic_bp"] / (df["diastolic_bp"] + 1e-3)

    # Additional signals
    df["pulse_pressure"] = df["systolic_bp"] - df["diastolic_bp"]
    df["mean_arterial_pressure"] = df["diastolic_bp"] + df["pulse_pressure"] / 3.0
    df["bmi_waist_ratio"] = df["bmi"] * df["waist_to_hip_ratio"]

    for col in ["cholesterol_total", "ldl_cholesterol", "triglycerides"]:
        df[f"log1p_{col}"] = np.log1p(df[col])

    return df

train_df_fe = add_interaction_features(train_df_fe)
test_df_fe = add_interaction_features(test_df_fe)

print("train_df_fe shape:", train_df_fe.shape)
print("test_df_fe shape :", test_df_fe.shape)

# Quick check: no missing values in features
print("\nRemaining missing values in train (top 10):")
display(train_df_fe.isnull().sum().sort_values(ascending=False).head(10))

print("\nRemaining missing values in test (top 10):")
display(test_df_fe.isnull().sum().sort_values(ascending=False).head(10))
```

Missing values were handled using median imputation for numerical features and a dedicated "Unknown" category for categorical variables. One observation with a missing target value was removed from the training set.
Several domain-informed interaction features were added to capture non-linear relationships between key health indicators, such as age and body mass index, physical activity and BMI, lipid balance, and blood pressure ratios. As a result, the dataset was expanded while remaining fully clean and suitable for downstream preprocessing and neural network training.
